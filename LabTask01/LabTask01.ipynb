### Part 7: Mapping and Transformation ###
1. Convert categorical values in the gender column to numeric codes.

2. Create a new column that categorizes age into groups (e.g., child, adult, senior).



```python
# 1

df['SexCode'] = df['Sex'].map({'male': 0, 'female': 1})
df[['Sex', 'SexCode']].head()
```

```python
# 2

def age_func(x):
    if pd.isnull(x):
        return None
    elif x <= 12:
        return 'Child'
    elif x <= 19:
        return 'Teenager'
    elif x <= 60:
        return 'Adult'
    else:
        return 'Senior'

df['AgeGroup'] = df['Age'].map(age_func)
df[['Age', 'AgeGroup']].head(10)
```

### Part 8: Python ###

1. Select the names column. (2.3)

2. Show only the rows where the age is above 60. (3.1)

3. Create a new column that converts fares from USD to Euros. (5.1)

```python
# Asked Claude "how to open csv file into a list without pandas" (but i chose the implementation that turns it into a dictionary bc it's cleaner :100: and probably easier

import urllib.request
import csv

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

with urllib.request.urlopen(url) as response:
    lines = [l.decode('utf-8') for l in response.readlines()]
    reader = csv.DictReader(lines)
    data = list(reader)
```

```python
%%time

# 1

names = [row['Name'] for row in data] # names 
print(names[:5]) # print up to 5
```

```python
%%time

# 2

# creating a list of elderly
elderly = [row for row in data if row['Age'] != '' and float(row['Age']) > 60]

# display
for row in elderly[:5]:
    print(row['Name'], '|', row['Age'])
```

```python
%%time

# 3

USD_TO_EUR = 0.86

for row in data:
    row['Fare_EUR'] = float(row['Fare']) * USD_TO_EUR

# check result
for row in data[:5]:
    print(row['Fare'], 'in Euros:', row['Fare_EUR'])
```
